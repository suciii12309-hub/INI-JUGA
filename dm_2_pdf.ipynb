```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
from IPython.display import display
import ipywidgets as widgets
import pandas as pd
import io
import base64

uploader = widgets.FileUpload(
    accept=".csv",
    multiple=False
)

display(uploader)
```

```python
file_info = uploader.value

if file_info:
    if isinstance(file_info, dict):
        nama_file = list(file_info.keys())[0]
        isi_file = file_info[nama_file]["content"]
    else:
        nama_file = file_info[0]["name"]
        isi_file = file_info[0]["content"]

    df = pd.read_csv(io.BytesIO(isi_file))

    print("Dataset berhasil diimpor!")
    print("Nama file:", nama_file)
    print("Ukuran dataset:", df.shape)

    display(df.head())
else:
    print("Silakan unggah file CSV terlebih dahulu.")
```

```python
# Menampilkan 5 data pertama
display(df.head())

# Menampilkan 5 data terakhir
display(df.tail())

# Menampilkan ukuran dataset
print("Jumlah baris:", df.shape[0])
print("Jumlah kolom:", df.shape[1])

# Menampilkan nama kolom
print("\nNama kolom:")
print(df.columns.tolist())

# Menampilkan informasi dataset
df.info()
```

```python
# Menghitung missing values
jumlah_missing = df.isnull().sum()

# Menghitung persentase missing values
persentase_missing = (
    df.isnull().sum() / len(df) * 100
).round(2)

# Membuat tabel ringkasan
tabel_missing = pd.DataFrame({
    "Jumlah Missing Values": jumlah_missing,
    "Persentase (%)": persentase_missing
})

display(tabel_missing)
```

```python
missing = df.isnull().sum()
missing = missing[missing > 0].sort_values(
    ascending=False
)

plt.figure(figsize=(10, 5))
missing.plot(kind="bar")

plt.title("Jumlah Missing Values per Kolom")
plt.xlabel("Nama Kolom")
plt.ylabel("Jumlah Data Kosong")
plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

```python
# Menghitung data duplikat
jumlah_duplikat = df.duplicated().sum()

print("Jumlah data duplikat:", jumlah_duplikat)

# Menampilkan data duplikat
display(df[df.duplicated()])
```

```python
# Memilih kolom numerik
kolom_numerik = df.select_dtypes(
    include=["number"]
).columns

for kolom in kolom_numerik:
    Q1 = df[kolom].quantile(0.25)
    Q3 = df[kolom].quantile(0.75)
    IQR = Q3 - Q1

    batas_bawah = Q1 - 1.5 * IQR
    batas_atas = Q3 + 1.5 * IQR

    outlier = df[
        (df[kolom] < batas_bawah) |
        (df[kolom] > batas_atas)
    ]

    print("\nKolom:", kolom)
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Batas bawah:", batas_bawah)
    print("Batas atas:", batas_atas)
    print("Jumlah outlier:", len(outlier))
```

```python
# Memilih kolom kategorik
kolom_kategorik = df.select_dtypes(
    include=["object", "category"]
).columns

for kolom in kolom_kategorik:
    print("\nKolom:", kolom)
    print("Jumlah nilai unik:",
          df[kolom].nunique(dropna=True))

    print("\n10 nilai yang paling sering muncul:")
    print(df[kolom].value_counts(
        dropna=False
    ).head(10))

    # Memeriksa spasi di awal dan akhir teks
    data_teks = df[kolom].dropna().astype(str)

    jumlah_spasi = (
        data_teks != data_teks.str.strip()
    ).sum()

    print("Jumlah teks dengan spasi berlebih:",
          jumlah_spasi)
```

```python
plt.figure(figsize=(7, 5))

df["type"].value_counts().plot(kind="bar")

plt.title("Distribusi Jenis Konten Netflix")
plt.xlabel("Jenis Konten")
plt.ylabel("Jumlah")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
# Membuat salinan dataset
df_bersih = df.copy()

print("Dataset asli:", df.shape)
print("Dataset untuk dibersihkan:", df_bersih.shape)
```

```python
# Mendeteksi nilai durasi yang salah masuk ke rating
mask = df_bersih["rating"].astype(
    "string"
).str.fullmatch(r"\d+\s*min", na=False)

# Memindahkan nilai durasi ke kolom duration
df_bersih.loc[mask, "duration"] = (
    df_bersih.loc[mask, "rating"]
)

# Mengosongkan rating yang salah
df_bersih.loc[mask, "rating"] = np.nan

print("Jumlah data yang diperbaiki:", mask.sum())

display(
    df_bersih.loc[
        mask, ["title", "rating", "duration"]
    ]
)
```

```python
for kolom in df_bersih.select_dtypes(
    include=["object"]
).columns:
    df_bersih[kolom] = df_bersih[kolom].apply(
        lambda x: x.strip()
        if isinstance(x, str)
        else x
    )

print("Pembersihan spasi selesai.")
```

```python
kolom_missing = [
    "director",
    "cast",
    "country",
    "rating",
    "duration",
    "date_added"
]

for kolom in kolom_missing:
    if kolom in df_bersih.columns:
        df_bersih[kolom] = (
            df_bersih[kolom].fillna("Tidak diketahui")
        )

print("Missing values pada kolom terpilih telah ditangani.")

display(
    pd.DataFrame({
        "Missing Sebelum": df.isnull().sum(),
        "Missing Sesudah": df_bersih.isnull().sum()
    })
)
```

```python
print("===== PERBANDINGAN DATASET =====")

print("Jumlah baris sebelum:", len(df))
print("Jumlah baris sesudah:", len(df_bersih))

print("\nJumlah kolom sebelum:", len(df.columns))
print("Jumlah kolom sesudah:", len(df_bersih.columns))

print("\nMissing Values Setelah Pembersihan:")
print(df_bersih.isnull().sum())

print("\nJumlah Duplikat Setelah Pembersihan:")
print(df_bersih.duplicated().sum())

display(df_bersih.head())
```

```python
# Menyimpan dataset di folder kerja saat ini
df_bersih.to_csv(
    "netflix_titles_bersih.csv",
    index=False
)

print("Dataset berhasil disimpan!")

# Memeriksa file yang tersimpan
import os

print(os.listdir("."))
```
